# Missing Data & Statistical Inference

This notebook presents the final method-within-mechanism results from the simulation project. The primary coefficient is $\beta_1=2$.


## Project design

The same incomplete dataset for each mechanism is supplied to all seven methods: Complete Case, Mean Imputation, Hot-Deck, Regression Imputation, Stochastic Regression, MICE and EM.


```python
import pandas as pd
import matplotlib.pyplot as plt

results = pd.read_csv("../results/master_results.csv")
mice = pd.read_csv("../results/mice_details.csv")
results
```

## MCAR

Compare all methods while holding MCAR fixed. The true value is $\beta_1=2$.


```python
mcar = results[results["Mechanism"] == "MCAR"].copy()
mcar[["Method","Beta1","SE","CI_Lower","CI_Upper","Bias","Absolute_Bias"]]
```

## MAR

Compare all methods while holding MAR fixed.


```python
mar = results[results["Mechanism"] == "MAR"].copy()
mar[["Method","Beta1","SE","CI_Lower","CI_Upper","Bias","Absolute_Bias"]]
```

## MNAR

Compare all methods while holding MNAR fixed. Under MNAR, missingness depends on the unobserved outcome, so MAR-based methods should not be interpreted as automatically solving the MNAR mechanism.


```python
mnar = results[results["Mechanism"] == "MNAR"].copy()
mnar[["Method","Beta1","SE","CI_Lower","CI_Upper","Bias","Absolute_Bias"]]
```

## MICE details

## Visual comparison


```python
methods = results["Method"].unique()
mechanisms = ["MCAR","MAR","MNAR"]
fig, ax = plt.subplots(figsize=(12,6))
for method in methods:
    sub = results[results["Method"] == method].set_index("Mechanism").reindex(mechanisms)
    ax.plot(mechanisms, sub["Beta1"], marker="o", label=method)
ax.axhline(2, linestyle="--", label="True beta1 = 2")
ax.set_xlabel("Missingness mechanism")
ax.set_ylabel("Beta1 estimate")
ax.set_title("Beta1 Estimates by Method and Missingness Mechanism")
ax.legend(bbox_to_anchor=(1.02,1), loc="upper left")
plt.tight_layout()
plt.show()
```

## Interpretation

The main numerical pattern is the downward distortion associated with mean imputation in this simulated dataset. The other methods are generally closer to the known coefficient of 2, but these are results from one simulation and should not be treated as universal rankings. The project therefore emphasizes method comparison within each missingness mechanism.
